```python
# Filter to just employees who left
leavers = df[df['Attrition'] == 'Yes']

# Number of people who left
num_leavers = leavers.shape[0]

# Their average monthly income, converted to annual
avg_annual_salary = leavers['MonthlyIncome'].mean() * 12

# Apply the 50% replacement-cost rule of thumb
estimated_cost = num_leavers * avg_annual_salary * 0.5

print(f"Employees who left: {num_leavers}")
print(f"Average annual salary of leavers: ${avg_annual_salary:,.0f}")
print(f"Estimated total cost of attrition: ${estimated_cost:,.0f}")
```
